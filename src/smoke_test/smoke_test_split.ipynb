dataset dummy

```python
records = []

n_products = 20
reviews_per_product = 3

review_counter = 1

for product_idx in range(1, n_products + 1):
    
    product_id = f"product_{product_idx:02d}"
    
    for review_idx in range(reviews_per_product):
        
        records.append({
            "review_id": f"review_{review_counter:03d}",
            "product_id": product_id,
            "timestamp": pd.Timestamp("2026-01-01") 
                         + pd.Timedelta(days=review_counter),
            "text": f"Dummy review {review_counter} for {product_id}",
        })
        
        review_counter += 1


dummy_df = pd.DataFrame(records)

print(dummy_df.head())

print("\nDataset shape:", dummy_df.shape)
print("Unique products:", dummy_df["product_id"].nunique())
print("Unique reviews :", dummy_df["review_id"].nunique())
```

```python
assert len(dummy_df) == 60
assert dummy_df["product_id"].nunique() == 20
assert dummy_df["review_id"].nunique() == 60

assert dummy_df["product_id"].isna().sum() == 0
assert dummy_df["review_id"].isna().sum() == 0

print("✅ Dummy dataset validation PASS")
```
